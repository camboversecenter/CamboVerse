# Wat Phnom Naga Staircase — 3D Gaussian Splatting on Kaggle

This CamboVerse training notebook turns the staircase capture into two open, web-ready artifacts:

- **Normal:** a small `.splat` for a ~$150 Android phone over 4G.
- **Ultra / VR:** a richer `.splat`; VR always uses Ultra fidelity.

Pipeline: **audit images → prepare frames → COLMAP poses → Nerfstudio `splatfacto` → evaluate → export PLY → convert to `.splat`**.

Source dataset: <https://www.kaggle.com/datasets/camboversecenter/staircase>

**Verified from the private Kaggle data card (7 October 2026):** version 1 contains one `stair/` directory with **421 JPEG photographs** totaling **1.32 GB**. Sampled files are **4096 × 3072**, landscape, and ordered by camera timestamp. The Kaggle license and provenance fields are still blank, so this notebook intentionally blocks publication until the Center confirms them.

```python
# 1. Configuration — edit provenance before publishing
from pathlib import Path

DATASET_URL = "https://www.kaggle.com/datasets/camboversecenter/staircase"
EXPECTED_INPUT = Path("/kaggle/input/staircase/stair")  # verified Kaggle directory
SITE = "wat-phnom-staircase"

# Required archive metadata. Keep UNCONFIRMED until the Center verifies each value.
CAPTURE_CREATOR = "UNCONFIRMED"
CAPTURE_DATE = "UNCONFIRMED"          # ISO date, e.g. 2026-10-01
PERMISSION_REFERENCE = "UNCONFIRMED"
CONTENT_LICENSE = "UNCONFIRMED"       # Intended: CC-BY-4.0 after approval

ITERATIONS = 30_000                     # production run
TARGET_FRAMES = 350                    # 421 source photos; bounded preview while retaining coverage
LONG_SIDE = 1_600                      # bounds COLMAP time and VRAM
MATCHING = "exhaustive"                # ordered still photos; safer than video-only sequential matching
MIN_REGISTERED_RATIO = 0.60
MIN_REGISTERED_IMAGES = 60

# 32 bytes per splat: about 5.8 MB Normal and 19.2 MB Ultra before HTTP compression.
NORMAL_SPLATS = 180_000
ULTRA_SPLATS = 600_000
MAX_SPLAT_AXIS = 0.10                  # removes giant sky/background floaters
MIN_ALPHA = 0.08

WORK = Path("/kaggle/working")
FRAMES = WORK / "frames"
PROC = WORK / "proc"
OUT = WORK / "out"
EXPORT = WORK / "export"
```

```python
# 2. Locate every image in the attached staircase dataset
import os
import re
from collections import Counter

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".webp", ".tif", ".tiff"}

def natural_key(path: Path):
    parts = re.split(r"(\d+)", path.as_posix().lower())
    return [int(part) if part.isdigit() else part for part in parts]

dataset_roots = [path for path in Path("/kaggle/input").iterdir() if path.is_dir()]
if EXPECTED_INPUT.exists():
    data_root = EXPECTED_INPUT
else:
    ranked = []
    for root in dataset_roots:
        count = sum(1 for path in root.rglob("*") if path.suffix.lower() in IMAGE_EXTENSIONS)
        ranked.append((count, root))
    assert ranked and max(ranked)[0] > 0, "No attached image dataset found. Use Kaggle's Add Input first."
    data_root = max(ranked)[1]

source_images = sorted(
    (path for path in data_root.rglob("*") if path.suffix.lower() in IMAGE_EXTENSIONS),
    key=natural_key,
)
assert source_images, f"No supported images found below {data_root}"

print("Dataset root:", data_root)
print("Images:", len(source_images))
print("Formats:", dict(Counter(path.suffix.lower() for path in source_images)))
print("First:", source_images[0].relative_to(data_root))
print("Last: ", source_images[-1].relative_to(data_root))
```

```python
# 3. Audit sharpness, exposure, dimensions, and exact duplicates
import hashlib
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image, ImageFile, ImageOps

ImageFile.LOAD_TRUNCATED_IMAGES = True  # tolerate partially-written JPEGs in the dataset

def image_audit(path: Path) -> dict:
    with Image.open(path) as original:
        rgb = ImageOps.exif_transpose(original).convert("RGB")
        width, height = rgb.size
        preview = rgb.copy()
        preview.thumbnail((512, 512), Image.Resampling.LANCZOS)
        gray = np.asarray(preview.convert("L"), dtype=np.float32)
        # Dependency-free focus proxy: variance of horizontal/vertical gradients.
        focus = float(np.var(np.diff(gray, axis=0)) + np.var(np.diff(gray, axis=1)))
        dark = float(np.mean(gray < 8))
        bright = float(np.mean(gray > 247))
    digest = hashlib.sha256(path.read_bytes()).hexdigest()
    return {
        "path": str(path), "name": path.name, "width": width, "height": height,
        "megapixels": width * height / 1_000_000, "focus_score": focus,
        "dark_clip": dark, "bright_clip": bright, "sha256": digest,
    }

audit = pd.DataFrame(image_audit(path) for path in source_images)
audit["exact_duplicate"] = audit.duplicated("sha256", keep="first")
audit.to_csv(WORK / "staircase-image-audit.csv", index=False)

display(audit[["name", "width", "height", "megapixels", "focus_score", "dark_clip", "bright_clip", "exact_duplicate"]].head(12))
print("Exact duplicates:", int(audit["exact_duplicate"].sum()))
print("Image sizes:", audit.groupby(["width", "height"]).size().sort_values(ascending=False).head(10).to_dict())
print("Focus score percentiles:", audit["focus_score"].quantile([0, .05, .5, .95, 1]).round(2).to_dict())
print("Strongly clipped images:", int(((audit.dark_clip > .20) | (audit.bright_clip > .20)).sum()))

# Contact sheet across the full capture order. Inspect for blur, people, lens changes,
# broken orbit continuity, and missing views of both naga rails and stair treads.
sample_indices = np.linspace(0, len(source_images) - 1, min(24, len(source_images))).round().astype(int)
fig, axes = plt.subplots(math.ceil(len(sample_indices) / 6), 6, figsize=(16, 3 * math.ceil(len(sample_indices) / 6)))
for axis in np.asarray(axes).reshape(-1):
    axis.axis("off")
for axis, index in zip(np.asarray(axes).reshape(-1), sample_indices):
    with Image.open(source_images[index]) as image:
        axis.imshow(ImageOps.exif_transpose(image).convert("RGB"))
    axis.set_title(f"{index}: {source_images[index].name}", fontsize=8)
    axis.axis("off")
plt.suptitle("Wat Phnom staircase capture — evenly sampled audit", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# 4. Prepare a bounded, ordered frame set for COLMAP
import shutil

# Exact duplicates add cost but no geometry. Keep all other frames by default;
# the audit table supports manual removal if identifiable people or bad frames exist.
usable = [Path(row.path) for row in audit.itertuples() if not row.exact_duplicate]
assert len(usable) >= MIN_REGISTERED_IMAGES, (
    f"Only {len(usable)} unique images; capture more overlapping views before training."
)

selected_indices = np.linspace(0, len(usable) - 1, min(TARGET_FRAMES, len(usable))).round().astype(int)
selected = [usable[index] for index in selected_indices]

shutil.rmtree(FRAMES, ignore_errors=True)
FRAMES.mkdir(parents=True)
for index, source in enumerate(selected):
    with Image.open(source) as original:
        image = ImageOps.exif_transpose(original).convert("RGB")
        image.thumbnail((LONG_SIDE, LONG_SIDE), Image.Resampling.LANCZOS)
        image.save(FRAMES / f"frame_{index:05d}.jpg", quality=95, optimize=True)

print(f"Prepared {len(selected)} / {len(usable)} unique images at <= {LONG_SIDE}px in {FRAMES}")
```

```python
# 5. Verify Kaggle GPU and record the software base
import platform
import subprocess
import torch

assert torch.cuda.is_available(), "Enable a GPU in Kaggle Notebook settings before continuing."
gpu_name = torch.cuda.get_device_name(0)
gpu_capability = torch.cuda.get_device_capability(0)
print("Python:", platform.python_version())
print("Torch:", torch.__version__, "| CUDA:", torch.version.cuda)
print("GPU:", gpu_name, "| capability:", gpu_capability)
subprocess.run(["nvidia-smi", "--query-gpu=name,memory.total", "--format=csv"], check=True)
```

```python
# 6. Install COLMAP, Nerfstudio, and the PLY reader
# Preserve Kaggle's working CUDA build by constraining its preinstalled Torch packages.
from importlib.metadata import PackageNotFoundError, version

torch_pins = []
for package in ("torch", "torchvision", "torchaudio"):
    try:
        torch_pins.append(f"{package}=={version(package)}")
    except PackageNotFoundError:
        pass
(WORK / "constraints.txt").write_text("\n".join(torch_pins) + "\n")

subprocess.run(["apt-get", "update", "-qq"], check=True)
subprocess.run(["apt-get", "install", "-y", "-qq", "colmap"], check=True)
subprocess.run([
    "pip", "install", "-q", "nerfstudio", "plyfile", "-c", str(WORK / "constraints.txt")
], check=True)

# PyTorch 2.6+ otherwise refuses Nerfstudio's own trusted checkpoint at export.
os.environ["TORCH_FORCE_NO_WEIGHTS_ONLY_LOAD"] = "1"
os.environ["MAX_JOBS"] = "1"  # CUDA extension compilation can exhaust RAM in parallel.
os.environ["TORCH_CUDA_ARCH_LIST"] = f"{gpu_capability[0]}.{gpu_capability[1]}"
os.environ["QT_QPA_PLATFORM"] = "offscreen"

print("Torch preserved:", torch_pins)
print("Nerfstudio:", version("nerfstudio"))
print("plyfile:", version("plyfile"))
```

```python
# 7. Estimate camera poses with COLMAP (the slowest stage)
process_command = [
    "ns-process-data", "images",
    "--data", str(FRAMES),
    "--output-dir", str(PROC),
    "--no-gpu",
    "--matching-method", MATCHING,
    "--num-downscales", "0",
]
print("Running:", " ".join(process_command))
subprocess.run(process_command, check=True)
```

```python
# 8. Registration gate — do not spend GPU hours on a broken camera solve
import json

transforms = json.loads((PROC / "transforms.json").read_text())
registered = len(transforms["frames"])
registered_ratio = registered / len(selected)
print(f"COLMAP registered {registered} / {len(selected)} images ({registered_ratio:.1%}).")

assert registered >= MIN_REGISTERED_IMAGES, (
    f"Only {registered} cameras registered; inspect blur/overlap and recapture or remove a disconnected orbit."
)
assert registered_ratio >= MIN_REGISTERED_RATIO, (
    f"Registration ratio {registered_ratio:.1%} is below {MIN_REGISTERED_RATIO:.0%}; fix COLMAP before training."
)
print("Registration gate passed.")
```

```python
# 9. Train Nerfstudio Splatfacto
train_command = [
    "ns-train", "splatfacto",
    "--data", str(PROC),
    "--output-dir", str(OUT),
    "--vis", "tensorboard",
    "--max-num-iterations", str(ITERATIONS),
]
print("Running:", " ".join(train_command))
subprocess.run(train_command, check=True)
```

```python
# 10. Evaluate the held-out views and export the Gaussian PLY
import glob

configs = sorted(OUT.rglob("config.yml"), key=lambda path: path.stat().st_mtime)
assert configs, "Training completed without a config.yml output."
config_path = configs[-1]
metrics_path = WORK / f"{SITE}-metrics.json"

evaluation = subprocess.run([
    "ns-eval", "--load-config", str(config_path), "--output-path", str(metrics_path)
])
if evaluation.returncode:
    print("WARNING: ns-eval did not complete. Keep the visual review mandatory and inspect the training log.")

subprocess.run([
    "ns-export", "gaussian-splat",
    "--load-config", str(config_path),
    "--output-dir", str(EXPORT),
], check=True)
ply_candidates = list(EXPORT.glob("*.ply"))
assert ply_candidates, "ns-export produced no PLY file."
ply_path = ply_candidates[0]
print("PLY:", ply_path, f"({ply_path.stat().st_size / 1e6:.1f} MB)")
```

```python
# 11. Convert PLY to the open antimatter15 .splat format
# Staircase-safe policy: no spherical centre crop (which can cut off long stairs).
# Rank valid, visible splats by opacity and projected volume, then keep tier budgets.
from plyfile import PlyData

SH_C0 = 0.28209479177387814

def ply_to_splat(ply_file: Path, output_file: Path, limit: int | None) -> dict:
    vertices = PlyData.read(ply_file)["vertex"].data
    xyz = np.stack([vertices["x"], vertices["y"], vertices["z"]], axis=1).astype(np.float32)
    scale = np.exp(np.stack([
        vertices["scale_0"], vertices["scale_1"], vertices["scale_2"]
    ], axis=1)).astype(np.float32)
    rotation = np.stack([
        vertices["rot_0"], vertices["rot_1"], vertices["rot_2"], vertices["rot_3"]
    ], axis=1).astype(np.float32)
    rotation /= np.linalg.norm(rotation, axis=1, keepdims=True) + 1e-12
    alpha = 1 / (1 + np.exp(-vertices["opacity"].astype(np.float32)))
    rgb = 0.5 + SH_C0 * np.stack([
        vertices["f_dc_0"], vertices["f_dc_1"], vertices["f_dc_2"]
    ], axis=1)

    valid = (
        np.isfinite(xyz).all(axis=1)
        & np.isfinite(scale).all(axis=1)
        & np.isfinite(rotation).all(axis=1)
        & (scale.max(axis=1) <= MAX_SPLAT_AXIS)
        & (alpha >= MIN_ALPHA)
    )
    importance = np.where(valid, alpha * np.cbrt(np.maximum(scale.prod(axis=1), 1e-18)), -1)
    order = np.argsort(-importance)[: int(valid.sum())]
    if limit is not None:
        order = order[:limit]

    packed = np.zeros(len(order), dtype=[
        ("position", "<f4", 3), ("scale", "<f4", 3),
        ("color", "u1", 4), ("rotation", "u1", 4),
    ])
    packed["position"] = xyz[order]
    packed["scale"] = scale[order]
    rgba = np.concatenate([rgb[order], alpha[order, None]], axis=1)
    packed["color"] = (np.clip(rgba, 0, 1) * 255).round().astype(np.uint8)
    packed["rotation"] = np.clip(rotation[order] * 128 + 128, 0, 255).astype(np.uint8)
    packed.tofile(output_file)
    return {
        "file": output_file.name,
        "splats": len(order),
        "bytes": output_file.stat().st_size,
        "source_splats": len(vertices),
    }

normal_path = WORK / f"{SITE}.splat"
ultra_path = WORK / f"{SITE}-ultra.splat"
normal_stats = ply_to_splat(ply_path, normal_path, NORMAL_SPLATS)
ultra_stats = ply_to_splat(ply_path, ultra_path, ULTRA_SPLATS)
print("Normal:", normal_stats)
print("Ultra / VR:", ultra_stats)
```

```python
# 12. Package outputs with provenance and reproducibility metadata
import datetime as dt

final_ply = WORK / f"{SITE}.ply"
shutil.copy2(ply_path, final_ply)

def sha256(path: Path) -> str:
    digest = hashlib.sha256()
    with path.open("rb") as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()

artifacts = [normal_path, ultra_path, final_ply, WORK / "staircase-image-audit.csv"]
if metrics_path.exists():
    artifacts.append(metrics_path)

manifest = {
    "title": "Wat Phnom Naga Staircase 3D Gaussian Splat",
    "created_utc": dt.datetime.now(dt.timezone.utc).isoformat(),
    "source_dataset": DATASET_URL,
    "capture_creator": CAPTURE_CREATOR,
    "capture_date": CAPTURE_DATE,
    "permission_reference": PERMISSION_REFERENCE,
    "content_license": CONTENT_LICENSE,
    "pipeline": "Nerfstudio splatfacto via COLMAP",
    "iterations": ITERATIONS,
    "input_images": len(source_images),
    "selected_images": len(selected),
    "registered_images": registered,
    "registered_ratio": registered_ratio,
    "normal": normal_stats,
    "ultra_vr": ultra_stats,
    "software": {
        "torch": torch.__version__, "cuda": torch.version.cuda,
        "nerfstudio": version("nerfstudio"), "gpu": gpu_name,
    },
    "files": {},
}
for artifact in artifacts:
    manifest["files"][artifact.name] = {
        "bytes": artifact.stat().st_size,
        "sha256": sha256(artifact),
    }

manifest_path = WORK / f"{SITE}-manifest.json"
manifest_path.write_text(json.dumps(manifest, indent=2) + "\n")
print(manifest_path.read_text())

unconfirmed = [
    key for key, value in {
        "capture_creator": CAPTURE_CREATOR,
        "capture_date": CAPTURE_DATE,
        "permission_reference": PERMISSION_REFERENCE,
        "content_license": CONTENT_LICENSE,
    }.items() if value == "UNCONFIRMED"
]
if unconfirmed:
    print("PUBLICATION BLOCKED until the Center confirms:", ", ".join(unconfirmed))
```

```python
# 13. Final acceptance checks
assert normal_path.stat().st_size == normal_stats["splats"] * 32
assert ultra_path.stat().st_size == ultra_stats["splats"] * 32
assert normal_stats["splats"] <= NORMAL_SPLATS
assert ultra_stats["splats"] <= ULTRA_SPLATS
assert ultra_stats["splats"] >= normal_stats["splats"]

print(f"Normal: {normal_path.name} — {normal_path.stat().st_size / 1e6:.1f} MB")
print(f"Ultra/VR: {ultra_path.name} — {ultra_path.stat().st_size / 1e6:.1f} MB")
print("Checks passed. Visual and device acceptance are still required.")
```
